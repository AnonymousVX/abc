```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten
from tensorflow.keras.regularizers import l2
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()

X_train = X_train / 255.0
X_test = X_test / 255.0

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
model1 = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

model1.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history1 = model1.fit(
    X_train,
    y_train,
    epochs=15,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
model2 = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation='relu', kernel_regularizer=l2(0.001)),
    Dense(64, activation='relu', kernel_regularizer=l2(0.001)),
    Dense(10, activation='softmax')
])

model2.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history2 = model2.fit(
    X_train,
    y_train,
    epochs=15,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
model3 = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation='relu', kernel_regularizer=l2(0.001)),
    Dropout(0.5),
    Dense(64, activation='relu', kernel_regularizer=l2(0.001)),
    Dropout(0.3),
    Dense(10, activation='softmax')
])

model3.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history3 = model3.fit(
    X_train,
    y_train,
    epochs=15,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
print("Without Regularization")
loss1, acc1 = model1.evaluate(X_test, y_test, verbose=0)
print("Loss:", loss1)
print("Accuracy:", acc1)

print("\nWith L2")
loss2, acc2 = model2.evaluate(X_test, y_test, verbose=0)
print("Loss:", loss2)
print("Accuracy:", acc2)

print("\nWith L2 + Dropout")
loss3, acc3 = model3.evaluate(X_test, y_test, verbose=0)
print("Loss:", loss3)
print("Accuracy:", acc3)

```

```python
plt.figure(figsize=(14, 10))

# 1. Training Accuracy
plt.subplot(2, 2, 1)

plt.plot(history1.history['accuracy'], label='No Regularization')
plt.plot(history2.history['accuracy'], label='L2')
plt.plot(history3.history['accuracy'], label='L2 + Dropout')

plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

# 2. Validation Accuracy
plt.subplot(2, 2, 2)

plt.plot(history1.history['val_accuracy'], label='No Regularization')
plt.plot(history2.history['val_accuracy'], label='L2')
plt.plot(history3.history['val_accuracy'], label='L2 + Dropout')

plt.title("Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

# 3. Training Loss
plt.subplot(2, 2, 3)

plt.plot(history1.history['loss'], label='No Regularization')
plt.plot(history2.history['loss'], label='L2')
plt.plot(history3.history['loss'], label='L2 + Dropout')

plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

# 4. Validation Loss
plt.subplot(2, 2, 4)

plt.plot(history1.history['val_loss'], label='No Regularization')
plt.plot(history2.history['val_loss'], label='L2')
plt.plot(history3.history['val_loss'], label='L2 + Dropout')

plt.title("Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

```python
print("="*55)
print("{:<25} {:<12} {:<12}".format("Model", "Loss", "Accuracy"))
print("="*55)

print("{:<25} {:<12.4f} {:<12.4f}".format(
    "Without Regularization", loss1, acc1))

print("{:<25} {:<12.4f} {:<12.4f}".format(
    "L2 Regularization", loss2, acc2))

print("{:<25} {:<12.4f} {:<12.4f}".format(
    "L2 + Dropout", loss3, acc3))
```
