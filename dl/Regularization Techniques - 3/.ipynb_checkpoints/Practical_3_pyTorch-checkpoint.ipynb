```python

pip install torch torchvision matplotlib

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt

from torch.utils.data import DataLoader, random_split

# 1. Device configuration
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)


# 2. Load MNIST dataset
transform = transforms.ToTensor()

full_dataset = torchvision.datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

test_dataset = torchvision.datasets.MNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)

# Split training data into training and validation
train_size = int(0.8 * len(full_dataset))
val_size = len(full_dataset) - train_size

train_dataset, val_dataset = random_split(
    full_dataset, [train_size, val_size]
)

train_loader = DataLoader(
    train_dataset, batch_size=64, shuffle=True
)

val_loader = DataLoader(
    val_dataset, batch_size=64, shuffle=False
)

test_loader = DataLoader(
    test_dataset, batch_size=64, shuffle=False
)

print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
print("Testing samples:", len(test_dataset))


# 3. Build MLP Model
class MLP(nn.Module):

    def __init__(self, dropout=False):
        super(MLP, self).__init__()

        self.network = nn.Sequential(
            nn.Flatten(),

            nn.Linear(784, 128),
            nn.ReLU(),

            # Dropout layer
            nn.Dropout(0.5) if dropout else nn.Identity(),

            nn.Linear(128, 64),
            nn.ReLU(),

            nn.Dropout(0.3) if dropout else nn.Identity(),

            nn.Linear(64, 10)
        )

    def forward(self, x):
        return self.network(x)


# 4. Evaluation Function
def evaluate_model(model, loader, criterion):

    model.eval()

    total_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            total_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    avg_loss = total_loss / len(loader)
    accuracy = 100 * correct / total

    return avg_loss, accuracy


# 5. Training Function
def train_model(model, name, l2=False, epochs=5):

    model = model.to(device)

    criterion = nn.CrossEntropyLoss()

    # Apply L2 regularization using weight_decay
    if l2:
        optimizer = optim.Adam(
            model.parameters(),
            lr=0.001,
            weight_decay=0.001
        )
    else:
        optimizer = optim.Adam(
            model.parameters(),
            lr=0.001
        )

    history = {
        "train_loss": [],
        "val_loss": [],
        "train_acc": [],
        "val_acc": []
    }

    for epoch in range(epochs):

        model.train()
        running_loss = 0

        # Training
        for images, labels in train_loader:

            images = images.to(device)
            labels = labels.to(device)

            optimizer.zero_grad()

            outputs = model(images)
            loss = criterion(outputs, labels)

            loss.backward()
            optimizer.step()

            running_loss += loss.item()

        train_loss = running_loss / len(train_loader)

        # Evaluate training and validation performance
        _, train_acc = evaluate_model(
            model, train_loader, criterion
        )

        val_loss, val_acc = evaluate_model(
            model, val_loader, criterion
        )

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(train_acc)
        history["val_acc"].append(val_acc)

        print(
            f"{name} | Epoch [{epoch+1}/{epochs}] "
            f"Train Loss: {train_loss:.4f} "
            f"Val Loss: {val_loss:.4f} "
            f"Train Acc: {train_acc:.2f}% "
            f"Val Acc: {val_acc:.2f}%"
        )

    return model, history


# 6. Train Model 1 - Without Regularization
print("\nTraining Model 1 - Without Regularization")

model1 = MLP(dropout=False)

model1, history1 = train_model(
    model1,
    "Without Regularization",
    l2=False
)


# 7. Train Model 2 - With L2 Regularization
print("\nTraining Model 2 - With L2 Regularization")

model2 = MLP(dropout=False)

model2, history2 = train_model(
    model2,
    "L2 Regularization",
    l2=True
)


# 8. Train Model 3 - L2 + Dropout
print("\nTraining Model 3 - L2 + Dropout")

model3 = MLP(dropout=True)

model3, history3 = train_model(
    model3,
    "L2 + Dropout",
    l2=True
)


# 9. Final Test Accuracy Comparison
print("\nFinal Test Results")

models = [
    ("Without Regularization", model1),
    ("L2 Regularization", model2),
    ("L2 + Dropout", model3)
]

for name, model in models:

    test_loss, test_acc = evaluate_model(
        model, test_loader, nn.CrossEntropyLoss()
    )

    print(
        f"{name}: "
        f"Test Loss = {test_loss:.4f}, "
        f"Test Accuracy = {test_acc:.2f}%"
    )


# 10. Plot Training and Validation Accuracy
plt.figure(figsize=(10, 5))

plt.plot(history1["train_acc"], label="Model 1 - Train")
plt.plot(history1["val_acc"], label="Model 1 - Validation")

plt.plot(history2["train_acc"], label="Model 2 - Train")
plt.plot(history2["val_acc"], label="Model 2 - Validation")

plt.plot(history3["train_acc"], label="Model 3 - Train")
plt.plot(history3["val_acc"], label="Model 3 - Validation")

plt.xlabel("Epochs")
plt.ylabel("Accuracy (%)")
plt.title("Training and Validation Accuracy")
plt.legend()
plt.grid()
plt.show()


# 11. Plot Training and Validation Loss
plt.figure(figsize=(10, 5))

plt.plot(history1["train_loss"], label="Model 1 - Train")
plt.plot(history1["val_loss"], label="Model 1 - Validation")

plt.plot(history2["train_loss"], label="Model 2 - Train")
plt.plot(history2["val_loss"], label="Model 2 - Validation")

plt.plot(history3["train_loss"], label="Model 3 - Train")
plt.plot(history3["val_loss"], label="Model 3 - Validation")

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.grid()
plt.show()
```
